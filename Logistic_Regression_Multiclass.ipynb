#### Logistic regression for multiclass classification

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import make_classification

X, y = make_classification(
    n_samples=1000,
    n_classes=3,
    n_features=4,
    n_informative=3,
    n_redundant=0,
    random_state=42
)
```

```python
X
```

```python
pd.DataFrame(X)
```

```python
y
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
# ovr- one vs rest
# suppose you want ot classify the data into categories such as cat dog and bird
# so ovr will do like below

    #          Logistic Regression
    #                 │
    #     ┌───────────┼───────────┐
    #     ↓           ↓           ↓
    #  Cat vs Rest  Dog vs Rest  Bird vs Rest

# Suppose a new animal comes in.

# The models might say:

# Cat model  → 0.20
# Dog model  → 0.85
# Bird model → 0.30

# Highest score is:# Dog = 0.85

# So the final prediction is:# Dog

# That's One-vs-Rest.

from sklearn.linear_model import LogisticRegression

model=LogisticRegression(multi_class='ovr')

model
```

```python
model.fit(X_train,y_train)
```

```python
y_pred=model.predict(X_test)

y_pred
```

```python
y_test
```

```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix

print(accuracy_score(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

```python

model=LogisticRegression(multi_class='multinomial')

model
```

```python
model.fit(X_train,y_train)
```

```python
y_pred=model.predict(X_test)

y_pred
```

```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix

print(accuracy_score(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

```python

model=LogisticRegression()

model
```

```python
# hyperparameter tuning

from sklearn.model_selection import GridSearchCV

params={
    'penalty':['l1','l2','elasticnet'],
    'C':[0.001,0.01,0.1,1,10],
    'multi_class':['ovr','multinomial'],
    "solver":['lbfgs','saga','liblinear','newton-cg']
}

grid=GridSearchCV(estimator=model,param_grid=params,cv=5,scoring="accuracy",n_jobs=-1)
```

```python
grid
```

```python
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python

y_pred=grid.predict(X_test)
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix

print(accuracy_score(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))
```
